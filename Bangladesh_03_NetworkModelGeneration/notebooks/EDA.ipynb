```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

all_data = []
for i in range(9): # Scenario 0 t/m 8
    filepath = f"../data/experiment/scenario{i}.csv"
    if os.path.exists(filepath):
        df = pd.read_csv(filepath)
        all_data.append(df)

# Combineer alles in één grote tabel
results_df = pd.concat(all_data, ignore_index=True)

# Boxplot
plt.figure(figsize=(12, 6))
sns.boxplot(data=results_df, x='Scenario', y='Avg_Travel_Time', palette="Set2")
sns.stripplot(data=results_df, x='Scenario', y='Avg_Travel_Time', color='black', alpha=0.5, jitter=True)

# Grafiek opmaken
plt.title('Average Truck Travel Time (Chittagong to Dhaka) across Scenarios', fontsize=14)
plt.xlabel('Scenario (0 = Baseline)', fontsize=12)
plt.ylabel('Average Travel Time (Minutes)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Horizontale lijn voor de ideale reistijd (ong. 360 minuten / 6 uur)
plt.axhline(y=360, color='r', linestyle=':', label='Ideal Travel Time (6 hrs)')
plt.legend()

plt.show()
```
